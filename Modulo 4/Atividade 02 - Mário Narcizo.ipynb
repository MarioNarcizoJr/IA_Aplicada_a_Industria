#Importações

```python
import cv2
from matplotlib import pyplot as plt
from deepface import DeepFace
```

#Leitura da Imagem

```python
caminho_img = "fla_campeao.jpg"
imagem_bgr = cv2.imread(caminho_img)
imagem_rgb = cv2.cvtColor(imagem_bgr, cv2.COLOR_BGR2RGB)

plt.imshow(imagem_rgb)
```

#Execução fo DeepFace

```python
resultado = DeepFace.analyze(
    img_path = imagem_rgb,
    actions = ['age', 'gender', 'emotion', 'race'],
    detector_backend  = 'opencv',
    enforce_detection = False
)

if isinstance(resultado, list) and len(resultado) > 0:
  resultado = resultado[0]
```

```python
#Imprimindo Resultado
print("\n\nResultado Geral da Análise\n")

#Idade Estimada
idade = resultado.get('age', 'desconhecida')
print(f'Idade Estimada: {idade}')

#Gênero Dominante
genero = resultado.get('dominant_gender', None)
print(f'Gênero Dominante: {genero}')

#Emoção Dominante
emocao_principal = resultado.get('dominant_emotion', 'desconhecida')
print(f'Emoção Dominante: {emocao_principal}')

#Etnia Dominante
raca_principal = resultado.get('dominant_race', 'desconhecida')
print(f'Raça Dominante: {raca_principal}')
```

#Desenhando retângulo nos rostos dos jogadores


```python
imagem = cv2.imread('fla_campeao.jpg')

face_cascade = cv2.CascadeClassifier('haarcascade_frontalface_default.xml')

imagem_cinza = cv2.cvtColor(imagem, cv2.COLOR_BGR2GRAY)

detector_de_rostos = face_cascade.detectMultiScale(imagem_cinza, 1.3, 5)

for(inicio_x, inicio_y, rosto_w, rosto_h) in detector_de_rostos:
  fim_x = inicio_x + rosto_w
  fim_y = inicio_y + rosto_h

  cv2.rectangle(
      imagem,
      pt1 = (inicio_x, inicio_y),
      pt2 = (fim_x, fim_y),
      color = (255,0 , 0),
      thickness=2
      )

imagem_rgb = cv2.cvtColor(imagem, cv2.COLOR_BGR2RGB)
plt.imshow(imagem_rgb)
```
