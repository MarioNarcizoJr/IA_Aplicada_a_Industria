```python
#Importa bibliotecas
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import warnings, requests, zipfile, io

#Ignora Warnings
warnings.filterwarnings("ignore")

#Importa função pra ler extensões arff
from scipy.io import arff
```

```python
##Baixando arquivos dataset
arquivo_url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00212/vertebral_column_data.zip"

r = requests.get(arquivo_url, stream=True)

arquivo_zip = zipfile.ZipFile(io.BytesIO(r.content))

arquivo_zip.extractall()
```

```python
#Lendo Arquivo
data = arff.loadarff('column_2C_weka.arff')

#Criando DataFrame
df = pd.DataFrame(data[0])

df.head()
```

```python
#Qtd de Linhas e Colunas
df.shape
```

```python
df.describe()
```

```python
%matplotlib inline

df.plot()
```

Este gráfico exibe as séries de cada atributo numérico. Pode indicar tendências ou variações gerais em todas as colunas.

```python
df.plot(kind='density', subplots=True, layout=(4,2), figsize=(12,12), sharex=False, sharey=False)

plt.show()
```

Os gráficos de densidade mostram a distribuição de cada variável numérica. Permitem visualizar a forma da distribuição, picos e assimetrias.

```python
df['degree_spondylolisthesis'].plot.density()
```

Este gráfico de densidade detalha a distribuição da variável 'degree_spondylolisthesis'. É possível observar a concentração e dispersão dos valores.

```python
df['degree_spondylolisthesis'].plot.hist()
```

O histograma ilustra a frequência de ocorrência dos valores para 'degree_spondylolisthesis'. Ajuda a entender a forma da distribuição e a identificar a moda.

```python
df['degree_spondylolisthesis'].plot.box()
```

O box plot de 'degree_spondylolisthesis' exibe a mediana, quartis e possíveis outliers. Revela a dispersão dos dados e a presença de valores extremos.

```python
df["class"].value_counts()
class_mapper = {b'Abnormal': 1, b'Normal': 0}

df["Class"] = df["class"].replace(class_mapper)
```

```python
df.plot.scatter(y='degree_spondylolisthesis', x='class')
```

Este gráfico de dispersão mostra a relação entre 'degree_spondylolisthesis' e a 'class'. Permite visualizar se há distinção nos níveis da variável numérica entre as classes.

```python
df.groupby("class").boxplot(
    fontsize=20,
    rot=90,
    figsize=(20,10),
    patch_artist=True
  )
```

Estes box plots comparam a distribuição de todas as variáveis numéricas entre as classes 'Abnormal' e 'Normal'. São úteis para identificar diferenças significativas na dispersão e mediana entre os grupos.

```python
pd.plotting.scatter_matrix(df, figsize=(15,15))
plt.show()
```

A matriz de dispersão exibe as relações bivariadas entre todos os pares de variáveis numéricas. Diagonalmente, mostra os histogramas ou distribuições de cada variável.

```python
matrix_info = df.corr(numeric_only=True)
matrix_info['Class'].sort_values(ascending=False)
fig, ax = plt.subplots(figsize=(10,10))
colormap = sns.color_palette("BrBG", 10)
sns.heatmap(matrix_info, cmap=colormap, annot=True, fmt=".2f")
plt.show()
```

O heatmap de correlação mostra a intensidade e direção das relações lineares entre as variáveis. Valores próximos a 1 ou -1 indicam forte correlação, enquanto 0 indica ausência de correlação linear.
